```python
# Remove rows where the target AQI value is missing
df = df.dropna(subset=['AQI'])
```

```python
# Check remaining missing values after removing missing AQI records
df.isnull().sum()
```

```python
# Separate input features from the target variable
x = df.drop(['AQI','AQI_Bucket','City','Date'], axis=1)
y = df['AQI']
```

```python
# Check the dimensions of the features and target
print("Features shape:", x.shape)
print("Target shape:",y.shape)
```

```python
# Split the data into training and testing sets
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(
    x,y, test_size=0.2, random_state=42
)
```

```python
# Create and train the Random Forest regression model
from sklearn.ensemble import RandomForestRegressor

model = RandomForestRegressor(n_estimators=100, random_state=42)

model.fit(x_train, y_train)
```

```python
# Predict AQI values for the test data
y_pred = model.predict(x_test)

# Display the first 10 predictions
print(y_pred[:10])
```

```python
# Evaluate the model performance
from sklearn.metrics import mean_absolute_error,r2_score

mae = mean_absolute_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

print("Mean Absolute Error:", mae)
print("R-squared Score:", r2)
```

```python
# Compare actual and predicted AQI values
comparison = pd.DataFrame({
    'Actual': y_test,
    'Predicted': y_pred
})

comparison.head(10)
```

```python
# Convert AQI value into an air quality category
def aqi_category(aqi):
  if aqi <= 50:
    return "Good"
  elif aqi <= 100:
    return "Satisfactory"
  elif aqi <= 200:
    return "Moderate"
  elif aqi <= 300:
    return "Poor"
  elif aqi <= 400:
    return "Very Poor"
  else:
    return "Severe"
```

```python
# Test the AQI category function with an example value
print(aqi_category(80))
```

```python
# Display the predicted AQI and its category
predicted_aqi = y_pred[0]

print("Predicted AQI:", predicted_aqi)
print("AQI Category:", aqi_category(predicted_aqi))
```

```python
def health_advisory(aqi):
  if aqi <= 50:
    return "Air Quality is good. Enjoy outdoor activities!"
  elif aqi <= 100:
    return "Air Quality is satisfactory. Sensitive people should be cautious!"
  elif aqi <= 200:
    return "People with respiratory problems should reducce prolonged outdoor activitiy"
  elif aqi <= 300:
    return "Avoid prolonged outdoor activity.Consider wearing a mask if necessary."
  elif aqi <= 400:
    return "Air Quality is very poor. Avoid prolonged outdoor activity."
  else:
    return "Air Quality is severe. Avoid outdoor activities and take necessary precautions."
```

```python
# Cell 25 — Save the model for deployment
import joblib

joblib.dump(model, 'aqi_model.pkl')
print("Model saved as aqi_model.pkl")
```

```python
# Cell 26 — Save the exact feature column order
feature_columns = list(x.columns)
joblib.dump(feature_columns, 'feature_columns.pkl')
print("Feature columns saved:", feature_columns)
```

```python
# Cell 27 — Combined predict function (this is your app's backend logic)
def predict_aqi(pollutant_dict):
    input_df = pd.DataFrame([pollutant_dict])[feature_columns]
    predicted_aqi = model.predict(input_df)[0]
    category = aqi_category(predicted_aqi)
    advisory = health_advisory(predicted_aqi)
    return {
        'predicted_aqi': round(float(predicted_aqi), 2),
        'category': category,
        'advisory': advisory
    }
```

```python
# Cell 28 — Test it
sample_input = x_test.iloc[0].to_dict()
print("Sample input:", sample_input)
result = predict_aqi(sample_input)
print("Prediction result:", result)
```
